```python
import pandas as pd
data = {
    'Roll No': [1, 1, 2, 2, 1, 2],
    'Subject': ['DS', 'Java', 'DS', 'Java', 'DS', 'Java'],
    'Marks': [25, 40, 30, 26, 40, 50],
    'Date': ['01-01-23', '02-01-23', '01-01-23', '02-01-23', '15-01-23', '15-01-23']
}

df = pd.DataFrame(data)

df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%y')

df = df.sort_values(
    by=['Roll No', 'Subject', 'Date'],
    ascending=[True, True, False]
)

result = []

for (roll_no, subject), group in df.groupby(['Roll No', 'Subject']):
    marks = list(group['Marks'])
    while len(marks) < 3:
        marks.append(0)
    marks = marks[:3]

    result.append({
        'Roll No': roll_no,
        'Subject': subject,
        'M1': marks[0],
        'M2': marks[1],
        'M3': marks[2],
        'Date': group.iloc[0]['Date'].strftime('%d-%m-%y')
    })

output_df = pd.DataFrame(result)

print(output_df)
```
